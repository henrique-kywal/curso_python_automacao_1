### Abrindo o Google Chrome

```python
import subprocess
import pyautogui
import time

link = "https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga"

subprocess.Popen(["open", "-a", "Google Chrome"])
```

### Carregando o site(link) numa nova aba

```python
time.sleep(5)
# Abre uma nova aba
pyautogui.hotkey("command", "t")

# Espera a aba abrir
time.sleep(2)

# Digita uma URL
pyautogui.typewrite(link)
pyautogui.press("enter")
```

```python
time.sleep(5)
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains

driver = webdriver.Chrome()

driver.get(link)
time.sleep(3)
campo = driver.find_element(By.XPATH, '//*[@id=":6"]/div/c-wiz/div/div/c-wiz[2]/div/div/section/div/table/tbody/tr[1]/td[1]/div/div')
time.sleep(1)

ActionChains(driver).double_click(campo).perform()
time.sleep(5)

campo = driver.find_element(By.XPATH, '//*[@id=":24"]/div/c-wiz/div/div/c-wiz[2]/div/div/section/div/table/tbody/tr/td[5]/div/div')
time.sleep(1)

from pathlib import Path

arquivo = Path("/Users/henriquekywal/Downloads/Vendas - Dez.xlsx")

if not arquivo.exists():
    print("Arquivo criado!")

else:
    print("Arquivo já existe!")
    arquivo.unlink()

ActionChains(driver).click(campo).perform()
time.sleep(5)


```

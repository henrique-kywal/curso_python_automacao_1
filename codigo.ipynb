### Abrindo o Google Chrome

```python
import pyautogui
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains

link = "https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga"

driver = webdriver.Chrome()

# Abre uma nova aba
driver.switch_to.new_window("tab")
time.sleep(2)
driver.get(link)
```

### Clicando no link para exportar o arquivo de dados

```python
time.sleep(5)

export_link = driver.find_element(By.XPATH, '//*[@id=":6"]/div/c-wiz/div/div/c-wiz[2]/div/div/section/div/table/tbody/tr[1]/td[1]/div/div')
time.sleep(1)

ActionChains(driver).double_click(export_link).perform()
```

### Clicando no link para baixar o arquivo de dados

```python
time.sleep(5)
download_link = driver.find_element(By.XPATH, '//*[@id=":24"]/div/c-wiz/div/div/c-wiz[2]/div/div/section/div/table/tbody/tr/td[5]/div/div')
time.sleep(1)

from pathlib import Path

data_file = Path.home() / "Downloads" / "Vendas - Dez.xlsx"
file = Path(data_file)

# Caso o arquivo não exista, não gera erro no Python.
file.unlink(missing_ok=True)

ActionChains(driver).click(download_link).perform()
time.sleep(5)

new_name = file.with_name("data_file.xlsx")

# Coloque (;) caso não queira que o Jupyter exiba o PosixPath ao apagar o arquivo.
file.rename(new_name);
```

### Abrir e ver o conteudo do arquivo baixado

```python
import pandas as pd

data_path = Path.home() / "Downloads" / "data_file.xlsx"

data_return = pd.read_excel(data_path)

display(data_return)

```

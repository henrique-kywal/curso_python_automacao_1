### Importando variáveis

```python
from dotenv import load_dotenv
import os

load_dotenv()

downloads_url = os.getenv("DOWNLOADS_URL")
print(downloads_url)
```

### Abrindo o Google Chrome

```python
import pyautogui
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains

link = "https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga"

driver = webdriver.Chrome()

# Abre uma nova aba
driver.switch_to.new_window("tab")
time.sleep(2)
driver.get(link)
```

### Clicando no botão exportar

```python
time.sleep(5)

campo = driver.find_element(By.XPATH, '//*[@id=":6"]/div/c-wiz/div/div/c-wiz[2]/div/div/section/div/table/tbody/tr[1]/td[1]/div/div')
time.sleep(1)

ActionChains(driver).double_click(campo).perform()
```

### Clicando para fazer download do arquivo

```python
time.sleep(5)
campo = driver.find_element(By.XPATH, '//*[@id=":24"]/div/c-wiz/div/div/c-wiz[2]/div/div/section/div/table/tbody/tr/td[5]/div/div')
time.sleep(1)

from pathlib import Path

arquivo = Path(downloads_url + "/Vendas - Dez.xlsx")

if not arquivo.exists():
    print("Arquivo criado!")

else:
    print("Arquivo já existe!")
    arquivo.unlink()

ActionChains(driver).click(campo).perform()
time.sleep(5)
print(arquivo)

from pathlib import Path

arquivo = Path.home() / "Downloads" / "Vendas - Dez.xlsx"
novo_nome = arquivo.with_name("meu_arquivo.xlsx")

arquivo.rename(novo_nome)
print(Path.home())
```
